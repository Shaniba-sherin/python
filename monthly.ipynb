```python
#logical an conceptual
marks={
    "anu":85,
    "rahul":72,
    "akhil":91,
    "meera":65,
    "neha":78
}
highest=max(marks,key=marks.get)
print(f"the highest mark is ",highest)
lowest=min(marks,key=marks.get)
print(f"lowest marks is ",lowest)
print("the average marks of students= ",key=marks.get/5)
marks.sort(key=lambda x:x>=75)
print(marks)
```
